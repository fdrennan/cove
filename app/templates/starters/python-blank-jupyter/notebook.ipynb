```python
import numpy as np
import matplotlib.pyplot as plt

x = np.random.default_rng(1).normal(size=1000)
print(f"mean {x.mean():.3f}, sd {x.std(ddof=1):.3f}")
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.hist(x, bins=40)
ax.set_title("1,000 draws from a normal distribution");
```
