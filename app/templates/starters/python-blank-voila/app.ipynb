A blank Voilà app: this notebook, run for each visitor, its code hidden and its widgets live.

```python
import ipywidgets as w
import numpy as np

out = w.Output()
points = w.IntSlider(value=200, min=10, max=1000, step=10, description="Points")

def show(change=None):
    x = np.random.default_rng(1).normal(size=points.value)
    with out:
        out.clear_output()
        print(f"{points.value} draws: mean {x.mean():.3f}, sd {x.std(ddof=1):.3f}")

points.observe(show, names="value")
show()
w.VBox([points, out])
```
